```python
from networkx.algorithms import clique
import time
start_time = time.time()

all_cliques = list(clique.find_cliques(G))


max_size = max(len(c) for c in all_cliques)
max_cliques = [c for c in all_cliques if len(c) == max_size]

end_time = time.time()
elapsed_time = end_time - start_time

print("maximum clique:", max_cliques)
print(f" temp: {elapsed_time:.6f} s")

```

```python

from networkx.algorithms import find_cliques
   
import time
start_time = time.time()
def find_first_clique(G, size):

    for clique_nodes in find_cliques(G):
        if len(clique_nodes) == size:
            return clique_nodes  
    return None

first_solution = find_first_clique(G, m)
end_time = time.time()
elapsed_time = end_time - start_time

print("first :", first_solution)
print(f"temp: {elapsed_time:.6f} s")

```
