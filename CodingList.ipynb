```python
class CodingList:
    @staticmethod
    def coding_list(n):
        """
        Retourne une liste de codes correspondant à la taille N.
        
        Traduction de :
        coding_list(N)->
           case N of 
              4 -> [8209,131361,66113,1153,16658,8738,132162,67714,33300,17444,10308,135300,1048,34856,20552,8328];
              ...
           end.
        """
        if n == 4:
            return [8209, 131361, 66113, 1153, 16658, 8738, 132162, 67714, 
                    33300, 17444, 10308, 135300, 1048, 34856, 20552, 8328]
        
        elif n == 5:
            return [131105, 8389697, 4196481, 2101505, 8705, 263202, 133186, 
                    8392834, 4202754, 2114050, 526372, 266308, 139396, 8405252, 
                    4227588, 1052712, 532552, 278664, 164104, 8454664, 8240, 
                    1065040, 557200, 327952, 131600]
        
        elif n == 6:
            return [2097217, 536875137, 268443905, 134234625, 67142657, 67585, 
                    4198466, 2105474, 536887554, 268468738, 134284290, 67241986, 
                    8396868, 4210820, 2130180, 536936964, 268567556, 134481924, 
                    16793672, 8421512, 4260104, 2228744, 537134088, 268961800, 
                    33587280, 16842896, 8519952, 4456976, 2622480, 537921552, 
                    65632, 33685664, 17039648, 8913440, 5243936, 2099232]
        
        elif n == 7:
            return [33554561, 34359755009, 17179902465, 8590001153, 4295100417, 
                    2147749889, 532481, 67125378, 33587458, 34359804418, 
                    17180001282, 8590198786, 4295495682, 2148540418, 134250628, 
                    67174660, 33686020, 34360001540, 17180395524, 8590987268, 
                    4297072644, 268501128, 134349064, 67371528, 34079752, 
                    34360789000, 17181970440, 8594137096, 537002128, 268697872, 
                    134742544, 68158480, 35653648, 34363936784, 17188266000, 
                    1074004128, 537395488, 269484576, 136315936, 71305248, 
                    41947168, 34376523808, 524480, 1074790720, 538968640, 
                    272630848, 142608448, 83890240, 33562688]
        
        elif n == 8:
            return [536871169, 2199023321601, 1099511759873, 549756078081, 
                    274878435329, 137440010241, 68721590273, 4227073, 
                    1073807618, 537002498, 2199023518722, 1099512154114, 
                    549756866562, 274880012290, 137443164162, 68727898114, 
                    2147614980, 1074004484, 537396228, 2199024306180, 
                    1099513729028, 549760016388, 274886311940, 137455763460, 
                    4295229704, 2148008456, 1074791432, 538970120, 2199027453960, 
                    1099520024584, 549772607496, 274911494152, 8590459152, 
                    4296016400, 2149581840, 1077938192, 545263632, 2199040040976, 
                    1099545198608, 549822955536, 17180918048, 8592032288, 
                    4299162656, 2155874336, 1090523168, 570433568, 2199090380832, 
                    1099645878304, 34361835840, 17184064064, 8598324288, 
                    4311746624, 2181042240, 1140858944, 671105088, 2199291723840, 
                    4194688, 34368127616, 17196647552, 8623491200, 4362080384, 
                    2281709696, 1342193792, 536903808]
        
        elif n == 9:
            return [8589935105, 140737488618497, 70368744704001, 35184373141505, 
                    17592188149761, 8796097232897, 4398054932481, 2199040098305, 
                    33685505, 17180131842, 8590459906, 140737489405954, 
                    70368746278914, 35184376291330, 17592194449410, 8796109832194, 
                    4398080131074, 2199090495490, 34360263172, 17180918788, 
                    8592033796, 140737492553732, 70368752574468, 35184388882436, 
                    17592219631620, 8796160196612, 4398180859908, 68720525832, 
                    34361836552, 17184065544, 8598327304, 140737505140744, 
                    70368777748488, 35184439230472, 17592320327688, 8796361588744, 
                    137441051152, 68723672080, 34368129040, 17196650512, 
                    8623497232, 140737555480592, 70368878428176, 35184640589840, 
                    17592723046416, 274882101792, 137447343136, 68736256032, 
                    34393296928, 17246986272, 8724168736, 140737756823584, 
                    70369281114144, 35185445961760, 549764203072, 274894685248, 
                    137472510016, 68786589760, 34493964352, 17448321088, 
                    9126838336, 140738562162752, 70370891792448, 1099528405632, 
                    549789369472, 274945017984, 137573175424, 68987920512, 
                    34896625792, 18253643904, 10737483904, 140741783453824, 
                    33555200, 1099578737920, 549890033920, 275146346752, 
                    137975832832, 69793235200, 36507255040, 21474902272, 
                    8590065920]
        
        elif n == 10:
            return [137438954497, 9007199255791617, 4503599629471745, 
                    2251799817887745, 1125899915247617, 562949970231297, 
                    281475010330625, 140737555595265, 70368878657537, 268959745, 
                    274878956546, 137441052674, 9007199258939394, 4503599635767298, 
                    2251799830478850, 1125899940429826, 562950020595714, 
                    281475111059458, 140737757052930, 70369281572866, 549757912068, 
                    274882103300, 137447346180, 9007199271526404, 4503599660941316, 
                    2251799880826884, 1125900041125892, 562950221987844, 
                    281475513843716, 140738562621444, 1099515823112, 549764204552, 
                    274894688264, 137472516104, 9007199321866248, 4503599761621000, 
                    2251800082186248, 1125900443844616, 562951027425288, 
                    281477124718600, 2199031645200, 1099528407056, 549789372432, 
                    274945024016, 137573187600, 9007199523209232, 4503600164306960, 
                    2251800887558160, 1125902054588432, 562954248912912, 
                    4398063289376, 2199056812064, 1099578740768, 549890039840, 
                    275146358816, 137975857184, 9007200328548384, 4503601774985248, 
                    2251804108914720, 1125908497301536, 8796126577728, 
                    4398113622080, 2199157477440, 1099780071488, 550292701248, 
                    275951681600, 139586502720, 9007203549839424, 4503608217567296, 
                    2251816994078784, 17592253154432, 8796227242112, 4398314950784, 
                    2199560134784, 1100585386112, 551903330432, 279172939904, 
                    146029019264, 9007216434872448, 4503633987633280, 35184506307840, 
                    17592454482176, 8796629897472, 4399120261376, 2201170755840, 
                    1103806628096, 558345814272, 292057907456, 171798954240, 
                    9007267974742272, 268436992, 35184908962304, 17593259790848, 
                    8798240514560, 4402341495296, 2207613223424, 1116691563008, 
                    584115683840, 343597646336, 137439478272]
        
        else:
            raise ValueError(f"Taille N={n} non supportée. N doit être entre 4 et 10.")

    @staticmethod
    def get_list_size(n):
        """Retourne la taille de la liste pour un N donné"""
        if n == 4:
            return 16
        elif n == 5:
            return 25
        elif n == 6:
            return 36
        elif n == 7:
            return 49
        elif n == 8:
            return 64
        elif n == 9:
            return 81
        elif n == 10:
            return 100
        else:
            raise ValueError(f"Taille N={n} non supportée")

    @staticmethod
    def decode_bits(code, n):
        """
        Décode un code en une liste de bits pour une taille n.
        Utile pour comprendre la structure des codes.
        """
        # Convertir le code en binaire
        binary = bin(code)[2:]
        
        # Compléter avec des zéros à gauche si nécessaire
        # La longueur totale devrait être un multiple de n
        padded_length = ((len(binary) + n - 1) // n) * n
        binary = binary.zfill(padded_length)
        
        # Diviser en groupes de n bits
        chunks = [binary[i:i+n] for i in range(0, len(binary), n)]
        
        return chunks

    @staticmethod
    def display_codes(n):
        """Affiche tous les codes pour une taille n avec leur représentation binaire"""
        codes = CodingList.coding_list(n)
        print(f"Codes pour N={n} (total: {len(codes)} codes):")
        print("=" * 60)
        
        for i, code in enumerate(codes):
            # Limiter l'affichage pour N >= 7 (codes très grands)
            if n >= 7 and i >= 5:
                if i == 5:
                    print(f"... {len(codes) - 5} codes supplémentaires ...")
                continue
            
            binary_repr = bin(code)[2:]
            # Pour les grands nombres, afficher une version tronquée
            if len(binary_repr) > 64:
                binary_display = f"{binary_repr[:32]}...{binary_repr[-32:]}"
                binary_display += f" ({len(binary_repr)} bits)"
            else:
                binary_display = binary_repr
            
            print(f"Code {i:3d}: {code:20d} = 0b{binary_display}")
        
        print("=" * 60)


# Version alternative avec optimisation de mémoire
class CompactCodingList:
    # Stocker les listes dans un dictionnaire pour éviter la recréation
    _cache = {}
    
    @classmethod
    def coding_list(cls, n):
        """Version avec cache pour éviter de recréer les listes"""
        if n not in cls._cache:
            # Utiliser la version de base
            cls._cache[n] = CodingList.coding_list(n)
        return cls._cache[n].copy()  # Retourner une copie pour éviter la modification


# Fonction utilitaire pour tester
def test_coding_list():
    """Teste le module coding_list"""
    
    # Tester pour différentes valeurs de N
    for n in range(4, 11):
        try:
            codes = CodingList.coding_list(n)
            expected_size = CodingList.get_list_size(n)
            
            print(f"N={n}:")
            print(f"  Nombre de codes: {len(codes)} (attendu: {expected_size})")
            print(f"  Premier code: {codes[0] if codes else 'N/A'}")
            print(f"  Dernier code: {codes[-1] if codes else 'N/A'}")
            
            # Vérifier que tous les codes sont différents
            unique_codes = set(codes)
            if len(unique_codes) == len(codes):
                print("  ✓ Tous les codes sont uniques")
            else:
                print(f"  ✗ {len(codes) - len(unique_codes)} codes en double")
            
            # Afficher quelques codes en binaire pour N=4
            if n == 4:
                print("\n  Exemples de décodage (N=4):")
                for i in [0, 1, 2, 3]:
                    code = codes[i]
                    chunks = CodingList.decode_bits(code, n)
                    print(f"    Code {i}: {code} = {chunks}")
            
            print()
            
        except ValueError as e:
            print(f"Erreur pour N={n}: {e}")
            print()


if __name__ == "__main__":
    # Tester le module
    test_coding_list()
    
    # Afficher les codes pour N=4
    print("Affichage détaillé pour N=4:")
    CodingList.display_codes(4)
    
    # Exemple d'utilisation
    print("\nExemple d'utilisation:")
    n = 5
    codes = CodingList.coding_list(n)
    print(f"Premiers 5 codes pour N={n}: {codes[:5]}")
```
